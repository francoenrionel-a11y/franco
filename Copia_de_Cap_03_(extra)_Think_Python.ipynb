No usaremos nada nuevo: solo `def`, parámetros, `print`, operadores aritméticos, `math.pi` y `for` con `range`.

```python
import math
```

## El rectángulo

Recordemos las fórmulas:

* **Área** = base × altura
* **Perímetro** = 2 × (base + altura)

Vamos a escribir una función para cada una.

```python
def calcular_area_rectangulo(base, altura):
    print(base * altura)
```

```python
calcular_area_rectangulo(4, 5)
```

Fíjate que `calcular_area_rectangulo` tiene **dos parámetros**: `base` y `altura`. Cuando llamamos a la función con `calcular_area_rectangulo(4, 5)`, el argumento `4` se asigna a `base` y el `5` a `altura`.

Ahora la función para el perímetro.

```python
def calcular_perimetro_rectangulo(base, altura):
    print(2 * (base + altura))
```

```python
calcular_perimetro_rectangulo(4, 5)
```

## El círculo

Para el círculo necesitamos $\pi$, que ya conocemos del Capítulo 2 (`math.pi`).

* **Área** = π × radio²
* **Perímetro (circunferencia)** = 2 × π × radio

```python
def calcular_area_circulo(radio):
    print(math.pi * radio ** 2)
```

```python
calcular_area_circulo(3)
```

```python
def calcular_perimetro_circulo(radio):
    print(2 * math.pi * radio)
```

```python
calcular_perimetro_circulo(3)
```

## Funciones que llaman a otras funciones

Como vimos en el Capítulo 3 con `print_verse`, una función puede llamar a otras funciones. Vamos a crear una función que resuma toda la información de un rectángulo, reutilizando las funciones que ya escribimos.

```python
def resumen_rectangulo(base, altura):
    print(f"Para un rectángulo de base {base} y altura {altura} se tiene: ")
    print('Área:')
    calcular_area_rectangulo(base, altura)
    print('Perímetro:')
    calcular_perimetro_rectangulo(base, altura)
```

```python
resumen_rectangulo(6, 2)
```

## Repetición con for

Usando lo que aprendimos sobre `for` y `range`, podemos calcular el área de varios círculos, con radios del 1 al 5, sin repetir código.

```python
for radio in range(1, 6):
    print('Radio:', radio)
    calcular_area_circulo(radio)
    print()
```

### Ejercicio 2

Escribe dos funciones para el cuadrado, que reciban un solo parámetro, `lado`:

* `calcular_area_cuadrado(lado)`: muestra el área (lado × lado)
* `calcular_perimetro_cuadrado(lado)`: muestra el perímetro (4 × lado)

### Ejercicio 4

Usa un `for` con `range` para mostrar el perímetro de circunferencias con radios del 1 al 10, usando la función `calcular_perimetro_circulo` que ya escribimos.

```python
import math

def calcular_perimetro_circulo(radio):
    perimetro = 2 * math.pi * radio
    print(f"Radio {radio}: Perímetro = {perimetro:.2f}")

for radio in range(1, 11):
    calcular_perimetro_circulo(radio)
```
